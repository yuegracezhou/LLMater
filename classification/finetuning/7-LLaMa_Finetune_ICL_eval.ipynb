```python
import pandas as pd

random = pd.read_csv('/content/content/results/test_bace_LLAMA_8_random.csv')
scaffold = pd.read_csv('/content/content/results/test_bace_LLAMA_8_scaffold.csv')
```

```python
import pandas as pd

# Encode 'Yes' as 1 and 'No' as 0 in the 'pred' column
random['pred_encoded'] = random['pred'].map({'Yes': 1, 'No': 0})
scaffold['pred_encoded'] = scaffold['pred'].map({'Yes': 1, 'No': 0})
```

```python
results = []
results.append(['mode','accuracy','precision','recall','f1'])
results
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_test = random['class_label']
y_pred = random['pred_encoded']

# Combine y_test and y_pred into a temporary DataFrame to easily drop rows with NaNs
temp_df = pd.DataFrame({'y_test': y_test, 'y_pred': y_pred})
temp_df = temp_df.dropna()

# Update y_test and y_pred with the cleaned data
y_test = temp_df['y_test']
y_pred = temp_df['y_pred']



# Evaluate the predictions
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
results.append(['random',accuracy,precision,recall,f1])

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
```
