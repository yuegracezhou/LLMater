```python
import pandas as pd

random4 = pd.read_csv('/content/results_samplenum4/test_bace_gpt-o4_4_random.csv')
random7 = pd.read_csv('/content/results_samplenum7/test_bace_gpt-o4_4_random.csv')
scaffold4 = pd.read_csv('/content/results_samplenum4/test_bace_gpt-o4_4_scaffold.csv')
scaffold7 = pd.read_csv('/content/results_samplenum7/test_bace_gpt-o4_4_scaffold.csv')
```

```python
import pandas as pd

# Encode 'Yes' as 1 and 'No' as 0 in the 'pred' column
random4['pred_encoded'] = random4['pred'].map({'Yes': 1, 'No': 0})
random7['pred_encoded'] = random7['pred'].map({'Yes': 1, 'No': 0})
scaffold4['pred_encoded'] = scaffold4['pred'].map({'Yes': 1, 'No': 0})
scaffold7['pred_encoded'] = scaffold7['pred'].map({'Yes': 1, 'No': 0})
```

```python
results = []
results.append(['mode','accuracy','precision','recall','f1'])
results
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_test = random4['class_label']
y_pred = random4['pred_encoded']

# Combine y_test and y_pred into a temporary DataFrame to easily drop rows with NaNs
temp_df = pd.DataFrame({'y_test': y_test, 'y_pred': y_pred})
temp_df = temp_df.dropna()

# Update y_test and y_pred with the cleaned data
y_test = temp_df['y_test']
y_pred = temp_df['y_pred']



# Evaluate the predictions
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
results.append(['random4',accuracy,precision,recall,f1])

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
```

```python
result_zero_shot = pd.read_csv('/content/zeroshot/zero-shot/zero_shot_bace_gpt-o4.csv')
result_zero_shot
```

```python
# Use apply with a lambda function to split each string in the 'pred' column
result_zero_shot['pred_val'] = result_zero_shot['pred'].apply(lambda x: x.split(':')[-1] if isinstance(x, str) and ':' in x else x)

# Encode 'Yes' as 1 and 'No' as 0 in the 'pred_val' column
result_zero_shot['pred_encoded'] = result_zero_shot['pred_val'].map({'Yes': 1, ' Yes': 1, 'No': 0})

# Display the updated DataFrame
result_zero_shot
```

```python
ytest2 = result_zero_shot['class_label']
ypred2 = result_zero_shot['pred_encoded']
```

```python
accuracy = accuracy_score(ytest2, ypred2)
precision = precision_score(ytest2, ypred2)
recall = recall_score(ytest2, ypred2)
f1 = f1_score(ytest2, ypred2)
results.append(['zeroshot',accuracy,precision,recall,f1])
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
```
